```python
import networkx as nx
import numpy as np

e = []
with open("Facebook_eps0.5_i5.txt", 'r') as f:
    for line in f.readlines():
        v, w = line.split()
        e.append((int(v), int(w)))

g = nx.Graph(e)

print(g)
```

```python
import pickle

with open("../Data/Facebook/nx_adj.pkl", 'rb') as a:
    f = pickle.load(a)

print(f)
```

```python
ga = nx.adjacency_matrix(g).toarray()
fa = nx.adjacency_matrix(f).toarray()
```

```python
c = 0
for i in range(4038):
    for j in range(i+1, 4039):
        if ga[i][j] == 1 and fa[i][j] == 1:
            c += 1

print(c/15135)
```

```python
np.all(fa==fa.T)
```

```python
import os
directory = "result/Facebook"
files = [f for f in os.listdir(directory) if os.path.isfile(os.path.join(directory, f))]
for file in files:
    e = []
    with open(os.path.join(directory, file), 'rb') as f:
        for l in f.readlines():
            v, w = l.split()
            e.append((int(v), int(w)))
    g = nx.Graph(e)
    filename = file.split("_", 1)[1:][0][:-4]
    with open(os.path.join(directory, f"nx_adj_{filename}.pkl"), 'wb') as f:
        pickle.dump(g, f)
```
