```python
import json
import pickle
import numpy as np

with open("musae_git_features.json", 'r') as f:
    d = json.load(f)
```

```python
a = [0 for _ in range(4004+1)]
f = 4004+1
n = len(d)

M = np.zeros((n,f))

for v, a_list in d.items():
    for a in a_list:
        M[int(v)][a] = 1

M
```

```python
with open("feature_matrix_full.pkl", 'wb') as f:
    pickle.dump(M.astype(int), f)
```

```python
from scipy.stats import entropy

def entropy_per_column(X):
    # Assumes binary features: 0 or 1
    probs = np.mean(X, axis=0)  # P(X=1) per feature
    ent = -probs * np.log2(probs + 1e-10) - (1 - probs) * np.log2(1 - probs + 1e-10)
    return ent

ent = entropy_per_column(M)
sorted_indices = np.argsort(-ent)  # descending entropy
print(sorted_indices)
```

```python
M = M[:, sorted_indices[:50]]
```

```python
with open("feature_matrix_50_highest_entropy.pkl", 'wb') as f:
    pickle.dump(M.astype(int), f)
```
